```python
students = {     "Rahul": 91,     "Ayesha": 84,     "Kiran": 67,     "Ali": 45 }
countp=0
countf=0
for name, marks in students.items():
    print(f"Name:{name},Marks:{marks}")
    if marks<=40:
        print("Fail")
        countf+=1
    else:
        print("Pass")
        countp+=1
print("no of students passed:",countp)
print("no of students failed:",countf)
```

```python
#prob-15
items = { "Milk":50, "Bread":40, "Eggs":70, "Rice":500 } 
bill = 0  
while True:
    item = input("Enter Item Name : ").title()
    if item.lower() == "done":
        break      
    if item in items:
        bill += items[item] 
        print(item, "Added")
    else: 
        print("Item Not Available")
print("Final Bill:",bill)
```

```python
#prob-19
student_info = {
    "Name": "Aarav",
    "Age": 21,
    "Course": "Computer Science",
    "City": "Mumbai"
}
print("--- Initial Details ---")
print("Dictionary Keys:", list(student_info.keys()))
print("Dictionary Values:", list(student_info.values()))
print("\nProfile Summary:")
for key, value in student_info.items():
    print(f"• {key}: {value}")
student_info.update({"City": "Bangalore"})
print("\n--- Updated Profile Summary ---")
for key, value in student_info.items():
    print(f"• {key}: {value}")
```

```python
#prob-20
systems = {
    "Engine": "OK",
    "Fuel": "OK",
    "Navigation": "FAIL",
    "Communication": "OK"
}
mission_failed = False
for system, status in systems.items():
    if status == "FAIL":
        print(f"System Failure: {system}")
        with open("mission_log.txt", "w") as file:
            file.write(f"Failure in {system} system.")
            
        print("Mission Aborted.")
        mission_failed = True
        break  
if not mission_failed:
    with open("mission_log.txt", "w") as file:
        file.write("Launch Successful")
    print("All systems clear. Launch Successful!")
```
