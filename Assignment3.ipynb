```python
#prob-4
password = input("Enter Password: ")
upper = 0
lower = 0
digits = 0
for ch in password:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1
    elif ch.isdigit():
        digits += 1
print("Uppercase Letters:", upper)
print("Lowercase Letters:", lower)
print("Digits:", digits)
if len(password) >= 8 and upper > 0 and lower > 0 and digits > 0:
    print("Strong Password")
else:
    print("Weak Password")
```

```python
#prob-7
def withdraw(balance, amount):
    if amount <= balance:
        balance = balance - amount
        return balance
    else:
        return "Insufficient Balance"
balance = float(input("Enter Balance: "))
amount = float(input("Enter Withdrawal Amount: "))
result = withdraw(balance, amount)
print("Result:", result)
```

```python
#prob-8
def calculate_grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 75:
        return "B"
    elif mark >= 60:
        return "C"
    else:
        return "F"
mark = int(input("Enter Student Marks: "))
grade = calculate_grade(mark)
print("Grade:", grade)
```

```python
#prob-9
def calculate_bill(price, quantity):#9
    total = price * quantity
    return total
price = float(input("Enter Price of Item: "))
quantity = int(input("Enter Quantity: "))
bill = calculate_bill(price, quantity)
print("\n----- BILL -----")
print("Price per Item : ₹", price)
print("Quantity       :", quantity)
print("Total Bill     : ₹", bill)
```

```python
#prob-10
discount = lambda price: price * 0.9
price = float(input("Enter Product Price: "))
final_price = discount(price)
print("Original Price:", price)
print("Price after 10% Discount:", final_price)
```

```python
#prob-11
area = lambda length, width: length * width
length = float(input("Enter Length: "))
width = float(input("Enter Width: "))
result = area(length, width)
print("Area of Rectangle:", result)
```
