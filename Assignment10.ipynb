```python
#prob-1
import pandas as pd

employees_data = pd.DataFrame({
    "EmployeeID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Rahul", "Ayesha", "Ali", "Kiran", "Priya",
             "Rohit", "Sneha", "Arjun", "Meena", "John"],
    "Department": ["HR", "IT", "Sales", "IT", "Finance",
                   "HR", "IT", "Sales", "Finance", "IT"],
    "City": ["Hyderabad", "Hyderabad", "Delhi", "Bangalore", "Chennai",
             "Hyderabad", "Hyderabad", "Delhi", "Kolkata", "Bangalore"],
    "Salary": [35000, 55000, 42000, 62000, 48000,
               39000, 57000, 65000, 51000, 72000],
    "Experience": [2, 4, 3, 5, 4, 3, 2, 6, 5, 7]
})

employees_data.to_csv("employees.csv", index=False)

print("employees.csv created successfully")
```

```python
departments_data = pd.DataFrame({
    "Department": ["HR", "IT", "Finance", "Marketing"],
    "Manager": ["Anil", "Kavya", "Ramesh", "Priyanka"],
    "Budget": [250000, 600000, 400000, 300000]
})

departments_data.to_csv("departments.csv", index=False)

print("departments.csv created successfully")
```

```python
import pandas as pd

employees = pd.read_csv("employees.csv")
departments = pd.read_csv("departments.csv")

print("Employees Dataset:")
print(employees)

print("\nDepartments Dataset:")
print(departments)
```

```python
print(employees.groupby("Department")["Salary"].mean())
```

```python
print(employees.groupby("Department")["EmployeeID"].count())
```

```python
#prob-3
print(employees.groupby("Department")["Salary"].max())
```

```python
#prob-4
print(employees.groupby(["Department","City"]).count())
```

```python
#prob-6
print(
    employees.groupby("Department")["Salary"].agg(["min", "max", "mean", "sum"]))
```

```python
#prob-7
print(
    employees.groupby("Department")["Experience"].agg(["min", "max", "mean"])
)
```

```python
#prob-8
print(employees.groupby("Department").agg({"Salary":["mean","max"],
                                          "Experience":["mean","min"]}))
```

```python
#prob-9
print(
    employees.groupby("Department").agg(
        Employee_Count=("EmployeeID", "count"),
        Total_Salary=("Salary", "sum"),
        Highest_Experience=("Experience", "max")
    )
)
```

```python
#prob-10
inner_merge = pd.merge(
    employees,
    departments,
    on="Department",
    how="inner"
)
print(inner_merge[["Name","Budget"]])
```

```python
#prob-11
left_merge = pd.merge(
    employees,
    departments,
    on="Department",
    how="left"
)
print(left_merge[["Name","Department","Budget"]])

print("\nDepartments with Missing Manager Information:")
print(left_merge[left_merge["Manager"].isna()])
```

```python
right_merge = pd.merge(
    employees,
    departments,
    on="Department",
    how="right"
)

print(right_merge)

print("\nDepartments Without Employees:")
print(
    right_merge[
        right_merge["EmployeeID"].isna()
    ][["Department", "Manager", "Budget"]]
)
```

```python
outer_merge = pd.merge(
    employees,
    departments,
    on="Department",
    how="outer"
)

print(outer_merge)
```

```python
indicator_merge = pd.merge(
    employees,
    departments,
    on="Department",
    how="outer",
    indicator=True
)

print(
    indicator_merge[
        indicator_merge["_merge"] == "left_only"
    ]
)
```

```python
import pandas as pd

# Step 1: Read datasets
employees = pd.read_csv("employees.csv")
departments = pd.read_csv("departments.csv")

# Step 2: Merge employee and department data
final_df = pd.merge(
    employees,
    departments,
    on="Department",
    how="outer",
    indicator=True
)

# Step 3: Display average salary department-wise
print("Average Salary Department-wise:")
print(
    employees.groupby("Department")["Salary"].mean()
)

# Step 4: Display total employees in each department
print("\nTotal Employees Department-wise:")
print(
    employees.groupby("Department").size()
)

# Step 5: Display manager of every department
print("\nDepartment Managers:")
print(
    departments[["Department", "Manager"]]
)

# Step 6: Find department with highest average salary
avg_salary = employees.groupby("Department")["Salary"].mean()

print("\nDepartment with Highest Average Salary:")
print(avg_salary.idxmax())

print("Highest Average Salary:")
print(avg_salary.max())

# Step 7: Identify departments without employees
departments_without_employees = departments[
    ~departments["Department"].isin(
        employees["Department"]
    )
]

print("\nDepartments Without Employees:")
print(departments_without_employees)

# Step 8: Display final merged DataFrame
print("\nFinal Merged DataFrame:")
print(final_df)
```
